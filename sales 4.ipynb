```python
sales = [1200, 2500, 1800, 3000, 2200]
print("Sales Data:", sales)
```

```python
total_sales = sum(sales)
print("Total Sales:", total_sales)
```

```python
average_sales = total_sales / len(sales)
print("Average Sales:", average_sales)
```

```python
highest_sales = max(sales)
print("Highest Sale:",highest_sale)
```

```python
lowest_sales = min(sales)
print("Lowest Sale :",lowest_sales)
```

### Sales Data Analysis – Results

In this project, I analyzed five sales values using Python. I used the built-in functions `sum()`, `max()`, and `min()` to calculate the total, highest, and lowest sales. I also calculated the average sales by dividing the total sales by the number of sales values.

**Results:**

* Total Sales: 10,700
* Average Sales: 2,140
* Highest Sale: 3,000
* Lowest Sale: 1,200
